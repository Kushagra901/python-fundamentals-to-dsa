## 1. Classes and Objects in Python
- **Class**: A user-defined blueprint or template for creating objects. It bundles data (attributes) and behavior (methods) together.
- **Object**: A real-world instance created from a class blueprint. Each object has its own state and identity.

### Syntax:
```python
class ClassName:
    # class body
    attribute = "value"

obj = ClassName()  # Instantiation
```

````python
# 1. Defining a simple Student class
class Student:
    name = "karan"                  # Class attribute

# Creating object instances of Student
s1 = Student()                      # Instantiating object 1
print("Student 1 name:", s1.name)   # Output: karan

s2 = Student()                      # Instantiating object 2
print("Student 2 name:", s2.name)   # Output: karan


# 2. Defining a Car class with multiple attributes
class Car:
    color = "blue"                  # Color attribute
    brand = "mercedes"              # Brand attribute

# Creating an instance of Car
car1 = Car()                        # Instantiating car object
print("\nCar Color:", car1.color)   # Output: blue
print("Car Brand:", car1.brand)     # Output: mercedes
````

## 2. Constructors (`__init__`) & the `self` Parameter
- **`__init__()` Method**: A special constructor method automatically invoked whenever a new instance of the class is created.
- **`self` Parameter**: A reference to the current instance of the class. It allows each instance to access and bind its own attributes and methods.
- **Default Constructor**: Constructor with no parameters (except `self`), initializing default states.
- **Parameterized Constructor**: Constructor that accepts arguments to initialize instance-specific data.

````python
# 1. Demonstrating Default Constructor
class DefaultStudent:
    # Default constructor (no additional arguments beyond self)
    def __init__(self):
        print("Default constructor called: creating new student record")

# Instantiating object triggers __init__ automatically
s_default = DefaultStudent()


# 2. Demonstrating Parameterized Constructor
class Student:
    # Parameterized constructor accepting instance-specific values
    def __init__(self, fullname, marks):
        self.name = fullname        # Assigning instance attribute 'name'
        self.marks = marks          # Assigning instance attribute 'marks'
        print(f"Creating new student '{self.name}' in database...")

# Creating instance 1 with arguments
s1 = Student("karan", 97)
print("Student 1 Details:", s1.name, "-", s1.marks)

# Creating instance 2 with arguments
s2 = Student("arjun", 78)
print("Student 2 Details:", s2.name, "-", s2.marks)
````

## 3. Class Attributes vs. Instance Attributes & Precedence
- **Class Attributes**: Defined directly in the class body. They are shared across all instances of the class and occupy memory once per class.
- **Instance Attributes**: Defined inside methods (typically `__init__`) using `self.attribute`. They are unique to each object instance.
- **Precedence Rule**: If a class attribute and an instance attribute share the same name, **Instance Attribute takes precedence over Class Attribute** during object lookup (`instance.attr`).

````python
# Defining Student class with both class and instance attributes
class Student:
    college_name = "dps"            # Class attribute (shared by all students)
    name = "anonymous"              # Class attribute (default fallback name)

    def __init__(self, name, marks):
        self.name = name            # Instance attribute (unique per student)
        self.marks = marks          # Instance attribute (unique per student)
        print(f"Adding new student: {self.name}")

# Instantiating student object s1
s1 = Student("karan", 97)

# Instance attribute 'name' shadows class attribute 'name'
print("Instance attribute (s1.name):", s1.name)                     # Output: karan
print("Instance attribute (s1.marks):", s1.marks)                   # Output: 97

# Instantiating student object s2
s2 = Student("arjun", 88)
print("Instance attribute (s2.name):", s2.name)                     # Output: arjun
print("Instance attribute (s2.marks):", s2.marks)                   # Output: 88

# Accessing shared class attribute via Class name and object instances
print("\nShared Class Attribute (Student.college_name):", Student.college_name)  # Output: dps
print("Access via s1:", s1.college_name)                                         # Output: dps
print("Access via s2:", s2.college_name)                                         # Output: dps
````

## 4. Conceptual & Technical Interview Questions

### Q1: What is the purpose of the `self` parameter in Python class methods?
**Answer:**
- `self` represents the specific instance of the class upon which the method is called.
- When calling `s1.get_name()`, Python internally translates this call into `Student.get_name(s1)`.
- It enables distinct object instances to store, access, and modify their own independent state and attributes without colliding with other instances.

### Q2: What is the difference between Class Attributes and Instance Attributes?
**Answer:**
- **Class Attributes**: Bound to the class itself and shared among all instances. Defined inside the class but outside any methods. Modifying a class attribute via `ClassName.attr` affects all instances that do not override it.
- **Instance Attributes**: Bound to a specific instance via `self.attr`. Defined inside instance methods (like `__init__`). Each object maintains its own private copy in its `__dict__`.

### Q3: What happens when an instance attribute has the exact same name as a class attribute?
**Answer:**
- When an attribute is accessed via `obj.attribute`, Python first checks the object's instance dictionary (`obj.__dict__`).
- If found, it returns the instance attribute immediately, shadowing the class attribute.
- If not found on the instance, Python falls back to looking up the attribute on the class (`obj.__class__.__dict__`).
- Therefore, the **instance attribute always takes precedence** over the class attribute when referenced via an instance.

### Q4: How does `__init__` differ from a constructor in C++ or Java?
**Answer:**
- In Python, `__new__()` actually allocates and creates the object in memory, while `__init__()` acts as the initializer that configures initial attributes on the newly created instance.
- Python does not support traditional constructor overloading by method signature; only the last defined `__init__()` in a class definition takes effect (though flexible parameterization is achieved using default arguments, `*args`, or `**kwargs`).
