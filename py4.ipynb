## 1. Dictionaries in Python
A **dictionary** is an unordered, mutable collection that stores data in **`key : value`** pairs enclosed in curly braces `{}`.
- Keys must be immutable and unique.
- Values can be of any data type (strings, integers, booleans, lists, dictionaries).


```python
# Dictionary creation
info = {
    "key": "value",
    "name": "apnacollege",
    "learning": "python",
    "age": 35,
    "is_adult": True,
    "subject": ["python", "c++"]
}

print("Type of info:", type(info))
print("Accessing 'learning':", info["learning"])

# Modifying and adding key-value pairs
info["name"] = "kushagra"             # Modify existing value
info["surname"] = "srivastava"        # Add new key-value pair
print("Updated Dictionary:", info)

# Null / Empty Dictionary
null_dict = {}                        # Create null dictionary
null_dict["name"] = "kushagra"        # Assign key to null dict
print("Populated null_dict:", null_dict)
```

## 2. Nested Dictionaries
A nested dictionary is a dictionary inside another dictionary.


```python
# Nested dictionary example
student = {
    "name": "rahul",
    "Subject": {
        "phy": 45,
        "Chem": 98
    }
}

print("Complete student dictionary:", student)
print("Subject sub-dictionary:", student["Subject"])
print("Chemistry marks:", student["Subject"]["Chem"])
```

## 3. Essential Dictionary Methods
- `len(dict)`: Returns total number of key-value pairs.
- `.keys()`: Returns all keys.
- `.values()`: Returns all values.
- `.items()`: Returns all `(key, value)` pairs as tuples.
- `.update(new_dict)`: Inserts or updates key-value pairs.


```python
info = {
    "key": "value",
    "name": "apnacollege",
    "learning": "python",
    "age": 35,
    "is_adult": True,
    "subject": ["python", "c++"]
}

# 1. len(): Total key count
print("Length of info:", len(info))

# 2. keys(): List of all keys
print("Keys:", list(info.keys()))
print("Total number of keys:", len(list(info.keys())))

# 3. values(): List of all values
print("Values:", info.values())

# 4. items(): List of all (key, value) tuples
print("Items:", info.items())
pair = list(info.items())
print("First pair tuple:", pair[0])    # ('key', 'value')

# 5. update(): Updates existing keys or adds new ones
info.update({"city": "varanasi"})
print("After update city:", info)

new_dict = {"City": "varanasi"}
print("new_dict:", new_dict)
```

## 4. Safe Querying: `.get()` vs. Direct Indexing `dict[key]`
- `dict["key"]`: Raises a **`KeyError`** if the key does not exist, causing the program to crash.
- `dict.get("key")`: Returns **`None`** (or a default value) if the key is missing without raising an error.


```python
# Comparison: info["learning"] vs info.get("learning")
print("Direct bracket access:", info["learning"])     # Output: python
print("get() method access:", info.get("learning"))    # Output: python

# Querying a non-existent key:
# print(info["learning2"])  # Will raise KeyError and crash script execution!

# info.get("learning2") safely returns None:
print("Safe query for missing key:", info.get("learning2"))  # Output: None
```

## 5. Practice Questions (Beginner Level)


### Practice Question 1: Dictionary of Subjects and Marks
Create a dictionary of 3 subjects and their marks, then print total marks and average score.


```python
# Practice 1: Subjects and marks calculation
marks_dict = {
    "physics": 85,
    "chemistry": 90,
    "maths": 95
}
total = sum(marks_dict.values())
average = total / len(marks_dict)

print("Practice 1 - Marks dictionary:", marks_dict)
print("Total Marks:", total)
print("Average Score:", round(average, 2))
```

### Practice Question 2: Character Frequency Counter
Count the frequency of each character in the string `"kushagra"` using a dictionary.


```python
# Practice 2: Character frequency count using dictionary
word = "kushagra"
freq = {}

for char in word:
    freq[char] = freq.get(char, 0) + 1

print("Practice 2 - Character frequency:", freq)
```

## 6. Conceptual & Interview Questions

### Q1: How are Python dictionaries implemented internally?
**Answer:** Python dictionaries are implemented using a **hash table**. Keys are passed through a hash function `hash(key)` to compute an index in memory, providing average $O(1)$ time complexity for insertions, lookups, and deletions.

### Q2: What can and cannot be used as dictionary keys?
**Answer:** Only **immutable / hashable** types (such as `str`, `int`, `float`, `tuple` containing immutable items) can be dictionary keys. Mutable types like `list`, `dict`, or `set` cannot be used as keys because their hash value can change.

### Q3: Why is `.get()` preferred over `dict[key]` in production code?
**Answer:** `.get()` avoids `KeyError` exceptions when retrieving optional or dynamic keys, allowing the code to supply default values cleanly (e.g. `dict.get("key", 0)`).
