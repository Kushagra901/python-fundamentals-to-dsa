## 1. Dictionary Basics & Mutability
A dictionary is an **ordered (since Python 3.7)**, **mutable** collection of key-value pairs.
- Keys must be **immutable / hashable** types (`str`, `int`, `float`, `tuple`).
- Values can be of **any data type** (including lists, sets, or nested dictionaries).
- Duplicate keys are not allowed; assigning to an existing key overwrites its value.

```python
# Dictionary creation with heterogeneous value types
info = {
    "name": "kushagra",       # String key -> String value
    "subject": ["python", "c++", "java"],  # String key -> List value
    "topic": ("dict", "set"),             # String key -> Tuple value
    "age": 19,               # String key -> Integer value
    "is_adult": True,        # String key -> Boolean value
    "marks": 94.5            # String key -> Float value
}

# Accessing and modifying dictionary values
print("Initial info dictionary:", info)
print("Access name:", info["name"])

# In-place mutation (modifying existing key)
info["name"] = "kush"         # Reassigning existing key
info["surname"] = "srivastava" # Adding new key-value pair
print("Updated info dictionary:", info)

# Creating an empty dictionary and populating it dynamically
null_dict = {}                 # Empty dictionary literal
null_dict["name"] = "kushagra" # Assign key to null dict
print("Populated null_dict:", null_dict)
```

## 2. Nested Dictionaries
A nested dictionary is a dictionary contained as a value inside another dictionary, ideal for modeling hierarchical or structured data.

```python
# Nested dictionary example: student profile with nested subject marks
student = {
    "name": "rahul",
    "Subject": {
        "phy": 45,
        "Chem": 98,
        "math": 92
    }
}

print("Complete student dictionary:", student)
print("Subject sub-dictionary:", student["Subject"])
print("Chemistry marks:", student["Subject"]["Chem"])  # Chained key access
```

## 3. Essential Dictionary Methods
- `len(dict)`: Returns the total number of top-level key-value pairs.
- `.keys()`: Returns a `dict_keys` view of all keys (can be converted to `list`).
- `.values()`: Returns a `dict_values` view of all values.
- `.items()`: Returns a `dict_items` view of all `(key, value)` tuples.
- `.update(new_dict)`: Inserts or overwrites key-value pairs in-place from another dictionary or iterable.

```python
info = {
    "key": "value",
    "name": "apnacollege",
    "learning": "python",
    "age": 35,
    "is_adult": True,
    "subject": ["python", "c++"]
}

# 1. len(): Total key count
print("Length of info:", len(info))

# 2. keys(): Extract all keys
print("Keys view:", info.keys())
print("Keys as list:", list(info.keys()))
print("Total number of keys:", len(list(info.keys())))

# 3. values(): Extract all values
print("Values:", info.values())

# 4. items(): Extract all (key, value) pairs as tuples
print("Items:", info.items())
pair_list = list(info.items())
print("First pair tuple:", pair_list[0])    # Output: ('key', 'value')

# 5. update(): Merge another dictionary into existing one
info.update({"city": "varanasi", "age": 36})  # Adds 'city', updates 'age'
print("After update:", info)
```

## 4. Safe Querying: `.get()` vs. Direct Indexing `dict[key]`
- `dict["key"]`: Raises a **`KeyError`** if the key is missing, crashing script execution if unhandled.
- `dict.get("key", default=None)`: Safely retrieves the value. If the key is not present, returns `None` (or a specified fallback default) without raising an exception.

```python
# Comparison: info["learning"] vs info.get("learning")
print("Direct bracket access:", info["learning"])        # Output: python
print("get() method access:", info.get("learning"))       # Output: python

# Querying a non-existent key:
# print(info["learning2"])  # Direct index raises KeyError: 'learning2'

# info.get() safely returns None or a custom fallback value:
print("Safe query (missing key):", info.get("learning2"))             # Output: None
print("Safe query with fallback:", info.get("learning2", "Not Found")) # Output: Not Found
```

## 5. Sets in Python (Creation, Uniqueness & Mutability)
A **Set** is an **unordered collection of unique, immutable elements**.
- **Duplicates Ignored**: Any duplicate elements are automatically filtered out.
- **Mutability**: The set itself is mutable (can add/remove elements), but individual elements inside must be **immutable / hashable** (`int`, `float`, `str`, `tuple`). Mutable items like `list` or `dict` cannot be elements of a set.
- **Empty Set Syntax**: `{}` creates an empty dictionary. To create an empty set, you must use `set()`.

```python
# 1. Set creation: duplicate values are automatically deduplicated
collection = {1, 2, 2, 3, 4, "hello", "hello", 4}
print("Set contents (duplicates removed):", collection)
print("Set length (unique elements only):", len(collection))
print("Data type:", type(collection))

# 2. Critical distinction: empty dict vs empty set
empty_dict = {}          # Creates empty dict
empty_set = set()        # Creates empty set
print("Type of {}:", type(empty_dict))      # <class 'dict'>
print("Type of set():", type(empty_set))    # <class 'set'>
```

## 6. Essential Set Methods
- `.add(elem)`: Inserts a single element into the set.
- `.remove(elem)`: Removes the element; raises `KeyError` if element is not present.
- `.clear()`: Removes all elements, resulting in an empty `set()`.
- `.pop()`: Removes and returns an arbitrary element from the set.
- `.union(set2)`: Returns a new set containing all distinct elements from both sets ($A \cup B$).
- `.intersection(set2)`: Returns a new set containing only elements common to both sets ($A \cap B$).

```python
# 1. Creating an empty set and adding elements
collect = set()
collect.add(1)                 # Add integer
collect.add(2)                 # Add integer
collect.add("hello")           # Add string
collect.add((1, 23, 3))        # Add tuple (valid because tuple is immutable)
print("After adding elements:", collect)

# 2. Removing an element
collect.remove(2)              # Removes element 2
print("After remove(2):", collect)

# 3. Popping an arbitrary element
words_set = {"hello", "apna", "college", "python"}
popped_item = words_set.pop()  # Removes and returns arbitrary element
print("Popped item:", popped_item)
print("Remaining words_set:", words_set)

# 4. Clearing the set
collect.clear()
print("After clear():", collect)  # Output: set()

# 5. Mathematical Set Operations: Union and Intersection
set1 = {1, 2, 3}
set2 = {2, 3, 4}

union_set = set1.union(set2)                # Combines all unique elements: {1, 2, 3, 4}
intersection_set = set1.intersection(set2)  # Common elements only: {2, 3}

print("Set 1:", set1)
print("Set 2:", set2)
print("Union (set1 | set2):", union_set)
print("Intersection (set1 & set2):", intersection_set)
```

### Practice Question 6: Storing 9 and 9.0 as Distinct Values in Sets
In Python, integer `9` and float `9.0` compare equal (`9 == 9.0`) and generate the same hash code (`hash(9) == hash(9.0)`). Consequently, standard sets treat them as identical and keep only one value (`{9}`).
Demonstrate how to store both `9` and `9.0` as separate distinct values.

```python
# Standard set behavior: 9 and 9.0 collide due to value & hash equality
default_set = {9, 9.0}
print("Standard set {9, 9.0} deduplicates to:", default_set)

# Solution 1: Store one value as a string representation
string_approach = {9, "9.0"}
print("Solution 1 (String Cast):", string_approach)

# Solution 2: Store typed tuple pairs containing (type_name, value)
typed_tuple_approach = {
    ("float", 9.0),
    ("int", 9)
}
print("Solution 2 (Typed Tuples):", typed_tuple_approach)
```

## 8. Conceptual & Interview Questions

### Q1: How are Python dictionaries implemented internally?
**Answer:** Python dictionaries are implemented using a **hash table** (specifically a compact array of indices pointing to key-value entries since Python 3.6). Keys are passed through `hash(key)` to compute their bucket location, providing average **$O(1)$** time complexity for insertions, deletions, and lookups.

### Q2: What can and cannot be used as dictionary keys and set elements?
**Answer:** Only **immutable / hashable** types (such as `str`, `int`, `float`, `tuple` containing immutable items, and `frozenset`) can be dictionary keys or set elements. Mutable types like `list`, `dict`, or `set` cannot be used because their hash value can change during execution.

### Q3: Why is `.get()` preferred over `dict[key]` in production code?
**Answer:** `.get()` prevents unhandled `KeyError` exceptions when querying keys that may be optional or dynamically generated, allowing safe default value fallbacks (e.g. `dict.get("key", 0)`).

### Q4: How do Sets guarantee uniqueness and what is the lookup time complexity?
**Answer:** Sets in Python are implemented as hash tables without values (keys only). When inserting or querying an element `x`, Python computes `hash(x)` and performs an **$O(1)$ amortized** lookup to check for existing entries, making set membership testing (`x in my_set`) vastly faster than list linear search ($O(N)$).

### Q5: Why does `{9, 9.0}` evaluate to `{9}` in Python?
**Answer:** In Python, numerical equality `9 == 9.0` evaluates to `True`, and their hash values are identical (`hash(9) == hash(9.0) == 9`). Because set deduplication relies on both hash collision and equality check, Python recognizes `9.0` as already present and disregards it.
