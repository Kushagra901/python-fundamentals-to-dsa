## 1. Dictionaries in Python
A **dictionary** is an unordered, mutable collection of key-value pairs. 
- **Keys** must be immutable/hashable types (`str`, `int`, `tuple`).
- **Values** can be of any data type (`str`, `int`, `list`, `dict`, etc.).


```python
# 1. Dictionary creation
info = {
    "key": "value",
    "name": "apnacollege",
    "learning": "python",
    "age": 35,
    "is_adult": True,
    "subject": ["python", "c++"]
}

print("Dictionary contents:", info)
print("Type:", type(info))
print("Total number of keys:", len(info))

# 2. Accessing and modifying dictionary values
print("Name:", info["name"])          # Accessing value by key
info["name"] = "kushagra"             # Modifying existing key
info["surname"] = "srivastava"        # Adding a new key-value pair
print("Updated Dictionary:", info)

# 3. Creating an empty (null) dictionary
null_dict = {}
null_dict["username"] = "kushagra901"
print("Populated null_dict:", null_dict)
```

---
## 2. Nested Dictionaries
A dictionary can contain other dictionaries as values, creating hierarchical data structures.


```python
# Nested dictionary example
student_record = {
    "name": "rahul",
    "subject": {
        "physics": 45,
        "chemistry": 98,
        "mathematics": 92
    }
}

print("Complete student record:", student_record)
print("Nested subject dictionary:", student_record["subject"])
print("Chemistry score:", student_record["subject"]["chemistry"])
```

## 3. Essential Dictionary Methods
Python dictionaries provide built-in methods to inspect and update key-value pairs efficiently.


```python
info = {
    "key": "value",
    "name": "apnacollege",
    "learning": "python",
    "age": 35,
    "is_adult": True,
    "subject": ["python", "c++"]
}

# 1. keys(): Returns a view of all dictionary keys
all_keys = list(info.keys())
print("Keys:", all_keys)
print("Total keys count:", len(all_keys))

# 2. values(): Returns a view of all dictionary values
all_values = list(info.values())
print("Values:", all_values)

# 3. items(): Returns a view of all (key, value) pairs as tuples
all_items = list(info.items())
print("Items (key-value pairs):", all_items)
print("First item tuple:", all_items[0])

# 4. update(dict): Inserts or updates key-value pairs in-place
info.update({"city": "varanasi"})
print("After update city:", info)
```

## 4. Safe Querying: `.get()` vs. Direct Indexing `dict[key]`
- `dict["key"]`: If the key does not exist, raises a **`KeyError`**, stopping program execution.
- `dict.get("key", default=None)`: If the key does not exist, returns **`None`** (or a provided default) without crashing.


```python
# Safe vs. Unsafe key access
print("Safe access for existing key:", info.get("learning"))  # Returns 'python'

# Querying non-existent key
result = info.get("learning2")  # Returns None instead of throwing KeyError
print("Safe access for missing key 'learning2':", result)     # Output: None

# Direct bracket access on missing key would raise KeyError:
# info["learning2"]  -> Raises KeyError: 'learning2'
```

## 5. Beginner Practice Problems & Solutions


```python
# Practice 1: Create a dictionary of student marks and calculate total & average
marks_dict = {
    "physics": 85,
    "chemistry": 90,
    "maths": 95
}
total_marks = sum(marks_dict.values())
avg_marks = total_marks / len(marks_dict)
print("Practice 1 - Total:", total_marks, "| Average:", round(avg_marks, 2))

# Practice 2: Count frequency of characters in a word
word = "kushagra"
char_frequency = {}
for char in word:
    char_frequency[char] = char_frequency.get(char, 0) + 1
print("Practice 2 - Character frequency in 'kushagra':", char_frequency)

# Practice 3: Check if a key exists in a dictionary
target_key = "physics"
if target_key in marks_dict:
    print(f"Practice 3 - '{target_key}' exists with score {marks_dict[target_key]}")

# Practice 4: Merge two dictionaries
dict_a = {"x": 10, "y": 20}
dict_b = {"y": 30, "z": 40}
merged_dict = {**dict_a, **dict_b}  # dict_b overrides common keys
print("Practice 4 - Merged Dictionary:", merged_dict)
```

## 6. Interview Questions & Conceptual Insights

### Q1: How are Python dictionaries implemented internally?
**Answer:** Python dictionaries are implemented using a **hash table**. Each key is hashed using Python's built-in `hash()` function to compute an index in an underlying memory table. This provides $O(1)$ average time complexity for lookups, insertions, and deletions.

### Q2: What types can be used as dictionary keys?
**Answer:** Only **immutable (hashable)** types can be dictionary keys (such as `int`, `float`, `str`, `tuple` containing only immutable objects). Mutable objects like `list`, `dict`, or `set` cannot be keys because their content (and therefore hash value) can change.

### Q3: What is the advantage of using `.get()` over `dict[key]`?
**Answer:** `.get(key, default)` prevents unhandled `KeyError` exceptions when querying keys whose presence is not guaranteed, allowing clean fallback values without verbose `try...except` blocks.
