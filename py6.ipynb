## 1. Function Definition & Invocation
A **function** is a reusable block of code that performs a specific task. Functions eliminate code redundancy, improve modularity, and make programs easier to test and maintain.

### Syntax:
```python
def function_name(param1, param2):
    # function body
    return result
```

````python
# 1. Function definition that computes and returns the sum of two numbers
def cal_sum(a, b):
    return a + b                    # Returns the computed sum to caller

# Function invocation (call) and storing the returned value
sum_val = cal_sum(1, 2)             # Passes 1 and 2 as arguments
print("Sum:", sum_val)              # Output: Sum: 3

# 2. Function without parameters that performs an action
def print_hello():
    print("hello")

# Calling print_hello multiple times demonstrates code reusability
print_hello()
print_hello()
print_hello()
````

## 2. Parameters, Arguments & Default Values
- **Parameters**: Variables listed inside the parentheses in the function definition (placeholders).
- **Arguments**: Actual values sent to the function when it is invoked.
- **Default Parameters**: Parameters assigned a default fallback value. Default parameters must always follow non-default parameters.

````python
# 1. Function with multiple parameters: Average of 3 numbers
def avg_sum(a, b, c):
    return (a + b + c) / 3

avg_result = avg_sum(3, 67, 5)
print("Average:", avg_result)       # Output: Average: 25.0

# 2. Function with default parameters
def cal_prod(a=1, b=1):
    print("Product:", a * b)
    return a * b

# Invoking with both arguments overridden
cal_prod(4, 5)                      # Output: Product: 20

# Invoking with default fallback values
cal_prod()                          # Output: Product: 1
````

### Practice Question 1: Length of a List
Write a function `print_len(lst)` that takes a list as an argument and prints its length.

````python
cities = ["delhi", "gurgaon", "noida", "chennai"]

def print_len(lst):
    print("List length:", len(lst))

print_len(cities)                   # Output: List length: 4
````

### Practice Question 2: Print Elements on a Single Line
Write a function `print_list(lst)` that prints all elements of a list in a single line separated by spaces.

````python
heroes = ["thor", "iron man", "captain america"]

def print_list(lst):
    for item in lst:
        print(item, end=" ")
    print()                         # Newline after list traversal

print_list(heroes)                  # Output: thor iron man captain america 
````

### Practice Question 3: Factorial Calculation (Iterative Function)
Write a function `cal_fact(n)` that calculates and prints the factorial of a positive integer $n$.

````python
def cal_fact(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    print(f"Factorial of {n}:", fact)
    return fact

cal_fact(6)                         # Output: Factorial of 6: 720
````

### Practice Question 4: Currency Converter (USD to INR)
Write a function `converter(usd_val, rate=83)` to convert USD to INR.

````python
def converter(usd_val, rate=83):
    inr_val = usd_val * rate
    print(f"{usd_val} USD = {inr_val} INR")
    return inr_val

converter(45)                       # Output: 45 USD = 3735 INR
````

Write a function `check_even_odd(n)` that returns `"Even"` if $n$ is even, and `"Odd"` if $n$ is odd.

## 4. Recursion Fundamentals
**Recursion** is a programming technique where a function calls itself to solve smaller subproblems of the same problem.

### Key Requirements for Recursion:
1. **Base Case**: The termination condition that stops further recursive calls and prevents infinite recursion (`RecursionError: maximum recursion depth exceeded`).
2. **Recursive Step / Work**: The self-referencing call that progresses towards the base case with reduced problem size.

````python
# 1. Recursive countdown demonstration
def show(n):
    if n == 0:                      # 1. Base Case: Stop when n reaches 0
        return
    print(n)                        # 2. Work: print current value
    show(n - 1)                     # 3. Recursive Call: decrease n towards 0

print("Countdown from 5:")
show(5)                             # Prints 5, 4, 3, 2, 1

# 2. Recursive Factorial: n! = n * (n - 1)!
def fact(n):
    if n == 0 or n == 1:            # Base Case: 0! = 1 and 1! = 1
        return 1
    return n * fact(n - 1)          # Recursive Step

print("Factorial of 7 (recursive):", fact(7))  # Output: 5040
````

## 5. Practice Questions (Recursion)

### Practice Question 6: Recursive Sum of First $n$ Natural Numbers
Write a recursive function `sum_natural(n)` to calculate $1 + 2 + 3 + \dots + n$.

````python
def sum_natural(n):
    if n == 0:                      # Base Case: Sum of 0 natural numbers is 0
        return 0
    return n + sum_natural(n - 1)   # Recursive Step

print("Sum of first 5 natural numbers:", sum_natural(5))    # Output: 15
print("Sum of first 10 natural numbers:", sum_natural(10))  # Output: 55
````

### Practice Question 7: Recursive List Traversal
Write a recursive function `print_list_recursive(lst, idx=0)` that prints every element in a list.

````python
fruits = ["mango", "litchi", "apple", "banana"]

def print_list_recursive(lst, idx=0):
    if idx == len(lst):             # Base Case: Stop when index reaches list length
        return
    print(lst[idx])                 # Print current element
    print_list_recursive(lst, idx + 1)  # Recursive Step: advance index

print("Traversing list recursively:")
print_list_recursive(fruits)
````

## 6. Conceptual & Technical Interview Questions

### Q1: What is the semantic difference between Parameters and Arguments?
**Answer:** 
- **Parameters**: Variables defined in the function signature (e.g. `a, b` in `def cal_sum(a, b):`).
- **Arguments**: Concrete values passed into the function upon invocation (e.g. `1, 2` in `cal_sum(1, 2)`).

### Q2: What are Default Parameters, and what rule governs their order in function definitions?
**Answer:**
- Default parameters provide fallback values if the caller omits corresponding arguments (e.g. `def cal_prod(a=1, b=1):`).
- **Rule**: In Python, all non-default positional parameters **must precede** default parameters in the parameter list. Defining `def func(a=1, b):` raises a `SyntaxError: parameter without a default follows parameter with a default`.

### Q3: What is the difference between a function using `return` vs. `print()`?
**Answer:**
- `print()` outputs text to the standard output stream (`sys.stdout`) for human inspection and returns `None`.
- `return` terminates function execution and sends a computed value back to the caller for assignment, further arithmetic, or downstream processing.

### Q4: What happens if a recursive function lacks a Base Case?
**Answer:**
- Without a base case (or with an unreachable base case), the function repeatedly adds stack frames to the call stack until Python reaches its recursion limit (default is typically 1000 frames), throwing a `RecursionError: maximum recursion depth exceeded`.

### Q5: What is the call stack and time/space complexity of recursive factorial `fact(n)`?
**Answer:**
- **Time Complexity**: $O(N)$ since $N$ total recursive calls are made sequentially.
- **Space Complexity (Call Stack)**: $O(N)$ auxiliary stack space because $N$ activation records (stack frames) remain suspended in memory until the base case $n \le 1$ begins unwinding.
