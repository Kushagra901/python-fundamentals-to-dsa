## 1. Function Definition & Calling
A **function** is a block of code which only runs when it is called.
We define a function using the def keyword.


```python
# Function definition
def calSum(a, b):
    sum = a + b                    # Calculates the sum of a and b
    print(sum)                     # Prints the calculated sum

# Calling the function
calSum(5, 10)                      # Passing arguments 5 and 10 -> Output: 15
```

---
## 2. Reusing Functions with Multiple Arguments
Once defined, a function can be called repeatedly with different inputs.


```python
# Calling calSum with different inputs
calSum(2, 4)                       # Output: 6
calSum(12, 17)                     # Output: 29
```

## 3. Conceptual & Interview Questions

### Q1: What is the purpose of functions in Python?
**Answer:** Functions promote code reusability, eliminate redundancy, and make code modular and easier to maintain.

### Q2: What is the difference between parameters and arguments in calSum(a, b) and calSum(5, 10)?
**Answer:** 
-  and  in def calSum(a, b): are **parameters** (variable placeholders in the function definition).
- 5 and 10 in calSum(5, 10) are **arguments** (actual values passed into the function when calling it).
