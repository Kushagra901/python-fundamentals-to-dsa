## 1. String Creation & Escape Characters
Strings can be created using single `'...'`, double `"..."`, or triple `"""..."""` quotes.
Escape sequences like `\n` (newline) and `\t` (tab) format strings.


```python
# String creation
str1 = "this is a string"               # Double quotes
str2 = 'this is my name'               # Single quotes
str3 = """this is a string"""           # Triple quotes (supports multi-line)

str4 = "this is apna college's"        # Single quote inside double quotes

# Escape sequences
str5 = "this is a string .\nwe are learning python"  # \n creates a new line
str6 = "this is .\t my room"                         # \t creates a tab space

print("str5 with newline:")
print(str5)
print("\nstr6 with tab:")
print(str6)
```

---
## 2. String Operations: Concatenation & Length
- **Concatenation:** Combining two strings using the `+` operator.
- **`len()`:** Returns the total number of characters in the string.


```python
# String concatenation and length
str1 = "ksu"
str2 = "9safdsgfsdghjf"
sum_str = str1 + str2                  # Concatenates str1 and str2

print("Length of str1:", len(str1))    # 3
print("Concatenated string:", sum_str) # ksu9safdsgfsdghjf
print("Length of str2:", len(str2))    # 14
```

## 3. String Indexing & Slicing
- **Indexing:** Accessing individual characters starting from `0`.
- **Slicing:** Extracting a substring using `str[start : end]` (character at `end` index is **not** included).
- **Negative Slicing:** Counting indices from the end (right to left) starting from `-1`.


```python
# Indexing (starts from 0)
s = "apna college"
ch = s[1]                              # Character at index 1 is 'p'
print("Character at index 1:", ch)

# Positive slicing: str[start : end] (end index not included)
print("Slice [1:6]:", s[1:6])          # 'pna c' (indices 1 to 5)
print("Slice [1:len]:", s[1:len(s)])   # 'pna college'
print("Slice [2:]:", s[2:])            # 'na college' (from index 2 to the end)

# Negative slicing: counting from right (-1 is last character)
# "k u s h a g r a"
#  0 1 2 3 4 5 6 7
# -8-7-6-5-4-3-2-1
str1 = "kushagra"
print("Negative slice [-6:-1]:", str1[-6:-1])  # 'shagr' (from index -6 to -2)
```

## 4. Built-in String Functions & Methods
Because strings in Python are **immutable**, these methods return new strings and do not alter the original string.


```python
# String functions demonstration
s = "i am studying python"

# endswith(sub): checks if string ends with target substring
print("endswith 'on':", s.endswith("on"))             # True

# capitalize(): capitalizes the first character of the string
print("capitalize():", s.capitalize())                 # 'I am studying python'

# replace(old, new): replaces all occurrences of substring
print("replace 'o' with 'a':", s.replace("o", "a"))   # 'i am studying pythan'
print("replace 'python' with 'java':", s.replace("python", "java")) # 'i am studying java'

# find(sub): returns first index where substring is found (-1 if not found)
print("find 'o':", s.find("o"))                       # Index of first 'o'

# count(sub): counts occurrences of substring in the string
print("count 'am':", s.count("am"))                   # 1
```

## 5. Conditional Statements (`if`, `elif`, `else`)
Conditional statements control the execution flow based on conditions.
- Indentation (4 spaces) defines the code block inside each branch.
- In `if-elif-else`, the first branch that evaluates to `True` runs, and all subsequent branches are skipped.


```python
# Conditional statement: Traffic Light Example
light = "green"

if light == "red":
    print("stop")
elif light == "green":
    print("go")
elif light == "yellow":
    print("wait")
print("end of the code")

# Difference between multiple 'if' vs 'elif'
num = 5
if num > 2:
    print("greater than 2")
elif num > 3:
    print("greater than 3 (skipped because first if was true)")
```

---
## 6. Nested Conditionals (Nesting `if` inside `if`)
Placing an `if` statement inside another `if` block.


```python
# Nesting if example
age = 95

if age >= 18:
    if age >= 80:
        print("cannot drive (senior age limit)")
    else:
        print("can drive")
else:
    print("cannot drive (underage)")
```

## 8. Conceptual & Interview Questions

### Q1: Why are strings immutable in Python?
**Answer:** Strings cannot be changed in-place after creation. This guarantees thread safety, allows Python to cache hash values for fast dictionary lookups, and enables string interning (reusing identical string objects in memory).

### Q2: How does negative slicing work internally?
**Answer:** Python adds `len(string)` to negative indices. For `"kushagra"` (length 8), `[-6:-1]` translates to `[8-6 : 8-1]` = `[2 : 7]`, extracting characters from index 2 to 6 (`"shagr"`).

### Q3: What is the significance of indentation in Python?
**Answer:** Unlike C/C++/Java which use curly braces `{}` to define scope, Python uses whitespace indentation (standard is 4 spaces). Misaligned indentation raises an `IndentationError`.
