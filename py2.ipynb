## 1. String Creation & Escape Sequences
Strings are immutable sequences of characters enclosed in single, double, or triple quotes.
Escape characters: `\n` (newline), `\t` (horizontal tab).


```python
# 1. String creation variations
str1 = "this is a string"              # Double quotes
str2 = 'this is my name'              # Single quotes
str3 = """this is a multi-line string""" # Triple quotes

# 2. Escape characters
str_newline = "this is a string .\nwe are learning python"  # Newline
str_tab = "this is .\t my room"                            # Tab spacing

print(str_newline)
print(str_tab)

# 3. Concatenation and length
s1 = "kush"
s2 = "srivastava"
full_name = s1 + " " + s2              # String concatenation using '+'
print("Full Name:", full_name)
print("Length of full name:", len(full_name)) # len() returns total character count
```

## 2. String Indexing & Slicing
- **Zero-Based Indexing:** Positive indices start from `0` (left to right).
- **Negative Indexing:** Starts from `-1` (right to left).
- **Slicing Syntax:** `string[start : stop : step]` (the character at `stop` index is **not included**).


```python
# 1. Zero-based indexing
text = "apna college"
first_char = text[0]    # 'a'
second_char = text[1]   # 'p'
print("Character at index 1:", second_char)

# 2. Positive slicing [start : stop]
print("Slice [1:6]:", text[1:6])            # 'pna c' (indices 1, 2, 3, 4, 5)
print("Slice [1:len]:", text[1:len(text)])  # 'pna college'
print("Slice [2:]:", text[2:])              # 'na college' (from index 2 to end)
print("Slice [:4]:", text[:4])              # 'apna' (from start to index 3)

# 3. Negative slicing (indexing from right to left)
name = "kushagra"
# Indices: k(0/-8), u(1/-7), s(2/-6), h(3/-5), a(4/-4), g(5/-3), r(6/-2), a(7/-1)
print("Negative slice [-6:-1]:", name[-6:-1])  # 'shagr'
```

## 3. String Methods
String objects come equipped with powerful built-in inspection and transformation methods.
*(Note: Because strings are immutable, methods return a new string rather than modifying in-place).*


```python
sample = "i am studying python"

# 1. endswith(suffix): Checks if string terminates with target substring
print("Ends with 'on':", sample.endswith("on"))       # True

# 2. capitalize(): Capitalizes the first character of the string
capitalized = sample.capitalize()
print("Capitalized:", capitalized)                     # "I am studying python"

# 3. replace(old, new): Replaces occurrences of a substring
print("Replace 'o' with 'a':", sample.replace("o", "a"))
print("Replace 'python' with 'java':", sample.replace("python", "java"))

# 4. find(sub): Returns first index where substring is found (-1 if absent)
print("First index of 'o':", sample.find("o"))         # Returns index

# 5. count(sub): Returns frequency of substring in the string
print("Count of 'am':", sample.count("am"))            # 1
```

## 4. Conditional Statements (`if`, `elif`, `else`)
Conditional branching executes blocks of code based on whether a boolean expression evaluates to `True`.
Python uses **indentation** (4 spaces) to delimit code blocks instead of curly braces `{}`.


```python
# 1. Basic if-elif-else ladder (Traffic Light Example)
light = "green"

if light == "red":
    print("Action: STOP")
elif light == "green":
    print("Action: GO")
elif light == "yellow":
    print("Action: WAIT")
else:
    print("Action: Invalid signal")

# 2. Student Grading System
marks = 85

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
else:
    grade = "D"

print(f"Student Marks: {marks} -> Grade: {grade}")
```

## 5. Nested Conditionals
A nested conditional is an `if` statement placed inside another `if` or `else` block.


```python
# Nested if example: Driving eligibility and age check
driver_age = 25

if driver_age >= 18:
    if driver_age >= 80:
        print("Status: Senior citizen - requires special medical driving clearance")
    else:
        print("Status: Eligible to drive")
else:
    print("Status: Underage - cannot drive")
```

## 7. Interview Questions & Conceptual Insights

### Q1: Why are strings immutable in Python?
**Answer:** Immutability provides several critical benefits:
1. **Thread Safety:** Immutable objects can be shared safely across threads without synchronization locks.
2. **Hashability & Performance:** Because string contents cannot change, their hash value can be cached, making string lookups in dictionaries extremely fast.
3. **Memory Optimization (String Interning):** Python can reuse identical string literals in memory.

### Q2: How does Python evaluate negative slicing like `str[-6:-1]`?
**Answer:** Negative indices are converted internally using the formula `len(string) + index`. For a string of length 8 (e.g. `"kushagra"`), index `-6` becomes $8 - 6 = 2$ (`'s'`), and index `-1` becomes $8 - 1 = 7$ (`'a'`). The slice `[2:7]` extracts characters from index 2 up to index 6 inclusive (`"shagr"`).

### Q3: What is the difference between `elif` and multiple independent `if` statements?
**Answer:** 
- In an `if-elif-else` chain, once any condition evaluates to `True`, that block executes and Python **skips all subsequent branches**.
- With separate `if` statements, Python evaluates every condition independently, which can lead to multiple blocks executing unintentionally.
