## 1. First Python Program & Output
The `print()` function is a built-in Python function used to display messages or values to the console.


```python
# Built-in print function outputting text and numbers
print("hello")               # Output string "hello"
print("kushagra")            # Output string "kushagra"
print("hello , kushagra")    # Output concatenated string with comma
print(23)                    # Output integer 23
print(36)                    # Output integer 36
print(25 + 34)               # Evaluate arithmetic expression and output sum (59)
```

## 2. Variables & Dynamic Typing
In Python, a variable is a named memory location that stores a value. Python is **dynamically typed**, meaning you do not need to explicitly declare the data type.


```python
# Variable assignment with different data types
name = "kush"       # 'name' stores a string value in quotes
age = 23            # 'age' stores an integer value (whole number)
price = 25.99       # 'price' stores a floating point value (decimal number)
age2 = age          # 'age2' is assigned the current value of 'age' (23)

# Printing variable values
print(name)                                                  # Prints: kush
print(age)                                                   # Prints: 23
print(price)                                                 # Prints: 25.99
print("My name is :", name, ",", "My age is :", age)         # Formatted output
print(age2)                                                  # Prints: 23

# Type inspection using the type() function
print(type(name))   # <class 'str'> (string)
print(type(price))  # <class 'float'> (floating point number)
```

## 3. Data Types in Python
Python provides several standard built-in data types:
- `int`: Whole integer numbers (e.g. `23`, `-5`)
- `float`: Real numbers with decimal points (e.g. `25.99`, `3.14`)
- `str`: Text wrapped in single, double, or triple quotes
- `bool`: Truth values (`True` or `False`)
- `NoneType`: Represents the absence of a value (`None`)


```python
# Data type inspection
age = 23            # Integer
old = False         # Boolean (True / False)
a = None            # NoneType (represents absence of value)

print(type(age))    # Output: <class 'int'>
print(type(old))    # Output: <class 'bool'>
print(type(a))      # Output: <class 'NoneType'>
```

## 5. Operators in Python
Operators perform mathematical, logical, and relational operations on values (operands).


```python
# 1. Arithmetic Operators
a = 5
b = 2
print("Addition (a + b):", a + b)           # 7
print("Subtraction (a - b):", a - b)        # 3
print("Multiplication (a * b):", a * b)     # 10
print("Division (a / b):", a / b)           # 2.5 (always returns float)
print("Modulo / Remainder (a % b):", a % b) # 1
print("Power / Exponentiation (a ** b):", a ** b) # 25 (5^2)

# 2. Relational / Comparison Operators (Return boolean: True/False)
x = 50
y = 20
print("Equal to (x == y):", x == y)         # False
print("Not equal to (x != y):", x != y)     # True
print("Greater than (x > y):", x > y)       # True
print("Less than (x < y):", x < y)          # False
print("Greater or equal (x >= y):", x >= y) # True
print("Less or equal (x <= y):", x <= y)    # False

# 3. Assignment Operators (Shorthand operations)
num = 20
num += 5    # Equivalent to: num = num + 5 -> 25
num -= 3    # Equivalent to: num = num - 3 -> 22
num *= 2    # Equivalent to: num = num * 2 -> 44
num /= 4    # Equivalent to: num = num / 4 -> 11.0
num %= 3    # Equivalent to: num = num % 3 -> 2.0
num **= 3   # Equivalent to: num = num ** 3 -> 8.0
print("Final assigned value of num:", num)

# 4. Logical Operators (and, or, not)
p = True
q = False
print("Logical AND (p and q):", p and q)    # False (True only if BOTH are true)
print("Logical OR (p or q):", p or q)       # True  (True if AT LEAST ONE is true)
print("Logical NOT (not p):", not p)        # False (Inverts boolean state)
```

## 6. Type Conversion & Type Casting
- **Implicit Conversion:** Python automatically converts one data type into another without user intervention (e.g. promoting `int` to `float`).
- **Explicit Type Casting:** Manually converting a data type using built-in constructors like `int()`, `float()`, `str()`.


```python
# 1. Implicit Type Conversion (Automatic)
int_num = 2
float_num = 4.23
result = int_num + float_num  # Integer 2 is automatically promoted to float 2.0
print("Implicit conversion result:", result)  # Output: 6.23
print("Result type:", type(result))           # <class 'float'>

# 2. Explicit Type Casting (Manual)
str_val1 = "2"
str_val2 = "4.25"

cast_int = int(str_val1)      # Converts string "2" to integer 2
cast_float = float(str_val2)  # Converts string "4.25" to float 4.25
print("Casted sum:", cast_int + cast_float)  # Output: 6.25
```

## 7. User Input Handling (`input()`)
The built-in `input()` function reads a line of text from standard input and **always returns it as a string (`str`)**.
To perform calculations, we must explicitly cast it to `int` or `float`.


```python
# Simulated demonstration of user input handling
# Example user inputs
user_name = "kushagra"
user_age = int("23")         # Explicitly casting string to int
user_mark = float("99.99")   # Explicitly casting string to float

print("Welcome:", user_name)
print("Your age is:", user_age, "| Type:", type(user_age))
print("Your mark is:", user_mark, "| Type:", type(user_mark))
```

## 9. Interview Questions & Conceptual Insights

### Q1: What makes Python dynamically typed?
**Answer:** In Python, variable types are bound to values at runtime rather than fixed to variable names at compile time. A variable name is simply a reference (pointer) to an object in memory.

### Q2: What is the difference between `/` and `//` in Python?
**Answer:** 
- `/` is standard floating-point division (e.g. `5 / 2 = 2.5`).
- `//` is floor (integer) division, which returns the largest integer less than or equal to the quotient (e.g. `5 // 2 = 2`).

### Q3: Why does `input()` always return a string?
**Answer:** Since user input can contain any arbitrary characters (letters, numbers, punctuation), returning a string is the safest default representation. The programmer can then cast the string explicitly to the desired target type (`int()`, `float()`, etc.).

### Q4: How does short-circuit evaluation work with logical `and` / `or`?
**Answer:**
- For `A and B`: If `A` is `False`, Python immediately returns `False` without evaluating `B`.
- For `A or B`: If `A` is `True`, Python immediately returns `True` without evaluating `B`.
