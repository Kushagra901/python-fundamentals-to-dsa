## 1. First Python Program & `print()` Function
The built-in `print()` function outputs text, numbers, or expressions to the console. Whatever is inside quotes `""` will be printed as a string literal.


```python
# First python program
print("hello")               # Prints string "hello"
print("kushagra")            # Prints string "kushagra"

# The print() built-in function outputs whatever is inside quotes as text
print("hello , kushagra")    # Prints "hello , kushagra"

# Printing numbers and evaluating mathematical expressions
print(23)                    # Prints integer 23
print(36)                    # Prints integer 36
print(25 + 34)               # Evaluates 25 + 34 and prints 59
```

## 2. Variables & Dynamic Types
A **variable** is a named container for storing data values. In Python, variables are dynamically typed.


```python
# Variable declaration and assignment
name = "kush"       # 'name' stores a string value enclosed in double quotes
age = 23            # 'age' stores an integer value (whole number)
price = 25.99       # 'price' stores a floating point value (decimal number)
age2 = age          # 'age2' is assigned the value of 'age' (23)

# Printing variables
print(name)                                           # Output: kush
print(age)                                            # Output: 23
print(price)                                          # Output: 25.99
print("My name is :", name, ",", "My age is :", age)  # Output: My name is : kush , My age is : 23
print(age2)                                           # Output: 23

# type() function tells the data type of the variable
print(type(name))   # <class 'str'>
print(type(price))  # <class 'float'>
```

## 3. Core Data Types
Python has several built-in primitive data types:
- `int`: Integers / whole numbers
- `bool`: Booleans (`True` or `False`)
- `NoneType`: Represents the absence of a value (`None`)


```python
# Data types demonstration
age = 23            # Integer data type
old = False         # Boolean data type (True or False)
a = None            # NoneType represents no value / null

print(type(age))    # <class 'int'>
print(type(old))    # <class 'bool'>
print(type(a))      # <class 'NoneType'>
```

## 6. Operators in Python

### 6.1 Arithmetic Operators
Used for mathematical calculations: `+` (addition), `-` (subtraction), `*` (multiplication), `/` (division), `%` (modulo/remainder), `**` (power/exponentiation).


```python
# Arithmetic operators
a = 5
b = 2

sum_res = a + b     # Addition: 5 + 2 = 7
sub_res = a - b     # Subtraction: 5 - 2 = 3
mul_res = a * b     # Multiplication: 5 * 2 = 10
div_res = a / b     # Division: 5 / 2 = 2.5 (returns float)
mod_res = a % b     # Modulo (remainder): 5 % 2 = 1
power_res = a ** b  # Exponentiation (power): 5 ** 2 = 25

print("Addition:", sum_res)
print("Subtraction:", sub_res)
print("Multiplication:", mul_res)
print("Division:", div_res)
print("Modulo:", mod_res)
print("Power:", power_res)
```

### 6.2 Relational / Comparison Operators
Compare two values and return a boolean (`True` or `False`): `==`, `!=`, `>`, `<`, `>=`, `<=`.


```python
# Relational / Comparison operators
a = 50
b = 20

print("a == b:", a == b)  # False: 50 is not equal to 20
print("a != b:", a != b)  # True:  50 is not equal to 20
print("a > b:", a > b)    # True:  50 is greater than 20
print("a < b:", a < b)    # False: 50 is not less than 20
print("a >= b:", a >= b)  # True:  50 is greater than or equal to 20
print("a <= b:", a <= b)  # False: 50 is not less than or equal to 20
```

### 6.3 Assignment Operators
Used to assign and modify values in variables using shorthand notations: `+=`, `-=`, `*=`, `/=`, `%=`, `**=`.


```python
# Assignment operators
a = 20
print("Initial a:", a)

a += 5   # a = a + 5 -> 25
print("After a += 5:", a)

a -= 3   # a = a - 3 -> 22
print("After a -= 3:", a)

a *= 2   # a = a * 2 -> 44
print("After a *= 2:", a)

a /= 4   # a = a / 4 -> 11.0
print("After a /= 4:", a)

a %= 3   # a = a % 3 -> 2.0
print("After a %= 3:", a)

a **= 8  # a = a ** 8 -> 2.0 ** 8 = 256.0
print("After a **= 8:", a)
```

### 6.4 Logical Operators
Perform logical operations on boolean expressions:
- `and`: Returns `True` only if **both** operands are `True`.
- `or`: Returns `True` if **at least one** operand is `True`.
- `not`: Inverts the boolean value (`True` becomes `False`, `False` becomes `True`).


```python
# Logical operators
a = True
b = False

# and operator: returns True only when both conditions are True
print("a and b:", a and b)  # False

# or operator: returns True if at least one condition is True
print("a or b:", a or b)    # True

# not operator: inverts the result
print("not a:", not a)      # False
print("not b:", not b)      # True
```

## 7. Type Conversion vs. Type Casting
1. **Type Conversion (Automatic / Implicit):** Python automatically promotes values to the superior type (e.g. `int` + `float` promotes to `float`).
2. **Type Casting (Manual / Explicit):** Manually converting a data type using constructors like `int()`, `float()`, `str()`.


```python
# 1) Type conversion (automatically done by Python)
a = 2
b = 4.23
print("Automatic conversion sum:", a + b)  # 6.23 (float is superior value, so int is promoted)

# 2) Type casting (manually converting type)
# To type cast into integer: int("2") -> 2
# To type cast into float: float("4") -> 4.0
a = int("2")       # Converts string "2" to integer 2
a = float("4")     # Converts string "4" to float 4.0
b = 4.25
print("Casted sum:", a + b)  # 4.0 + 4.25 = 8.25
```

## 10. Conceptual & Interview Questions

### Q1: What does it mean that Python is dynamically typed?
**Answer:** In Python, variable types are determined and checked at runtime rather than at compile time. You don't need to specify `int a = 5`; writing `a = 5` dynamically assigns the integer type to `a`.

### Q2: What is the difference between `/` and `//` in Python?
**Answer:**
- `/` performs float division (e.g. `5 / 2 = 2.5`).
- `//` performs floor/integer division, discarding the fractional part (e.g. `5 // 2 = 2`).

### Q3: Why does `input()` always return a string?
**Answer:** Because user input can be any sequence of characters, strings are the most universal format. Python requires the programmer to explicitly cast (`int()`, `float()`) to avoid unintended type assumptions.

### Q4: How does short-circuit evaluation work with `and` / `or`?
**Answer:**
- In `A and B`, if `A` is `False`, Python immediately returns `False` without evaluating `B`.
- In `A or B`, if `A` is `True`, Python immediately returns `True` without evaluating `B`.
