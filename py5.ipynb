## 1. `while` Loop Fundamentals
A `while` loop repeatedly executes a block of code **as long as a specified boolean condition evaluates to `True`**.

### Anatomy of a `while` Loop:
1. **Initialization**: Set up a loop counter variable (e.g. `count = 1`).
2. **Condition**: Test the counter against a stopping threshold (e.g. `count <= 5`).
3. **Body**: Execute statements inside the loop.
4. **Update / Increment**: Modify the loop counter (e.g. `count += 1`) to guarantee eventual termination.

```python
# Basic while loop demonstration: print message 5 times
count = 1                          # 1. Initialize counter variable
while count <= 5:                  # 2. Condition check
    print("Hello World, iteration:", count)  # 3. Loop body
    count += 1                     # 4. Increment step (prevents infinite loop)

print("Loop completed. Final counter value:", count)  # Output: 6
```

## 2. Step Loops & Reverse Countdown Loops
Loops can decrement to count backward or step by arbitrary increments.

```python
# 1. Forward counter with custom step
i = 1
print("Forward counting (step = 2):")
while i <= 10:
    print(f"Step value: {i}")
    i += 2                         # Step by 2

# 2. Reverse countdown loop (decrementing)
countdown = 5
print("\nReverse countdown from 5 to 1:")
while countdown >= 1:
    print(f"T-minus: {countdown}")
    countdown -= 1                 # Decrement counter

print("Blastoff! Loop ended successfully.")
```

## 3. Traversing Sequences with Loops
We can traverse lists, tuples, or strings using an integer index variable.

```python
# Traversing a list using index pointer
heroes = ["Ironman", "Thor", "Captain America", "Hulk", "Hawkeye"]
idx = 0

print("Traversing list with while loop:")
while idx < len(heroes):
    print(f"Index {idx} -> Hero: {heroes[idx]}")
    idx += 1                       # Move to next element

# Traversing a string character by character
language = "PYTHON"
char_idx = 0
print("\nTraversing string:")
while char_idx < len(language):
    print(f"Char {char_idx}: {language[char_idx]}")
    char_idx += 1
```

## 4. Loop Flow Control: `break` and `continue`
- `break`: Immediately **terminates the entire loop** and transfers control to the statement directly following the loop.
- `continue`: Immediately **skips the rest of the current iteration** and jumps to the next iteration / condition check.

```python
# 1. Demonstration of 'break' (exit loop when number is 3)
print("--- Demonstrating 'break' ---")
n = 1
while n <= 5:
    if n == 3:
        print(f"Breaking early at n = {n}")
        break                      # Terminate loop prematurely
    print("Current n:", n)
    n += 1

# 2. Demonstration of 'continue' (skip printing even numbers)
print("\n--- Demonstrating 'continue' ---")
val = 0
while val < 6:
    val += 1
    if val % 2 == 0:               # Check if even
        continue                   # Skip even numbers
    print("Odd value:", val)
```

## 5. `for` Loops & the `range()` Function
A `for` loop iterates directly over elements of a sequence (list, tuple, string, or range).

### The `range()` Function:
- `range(stop)`: Generates numbers from `0` up to `stop - 1`.
- `range(start, stop)`: Generates numbers from `start` up to `stop - 1`.
- `range(start, stop, step)`: Generates numbers incrementing by `step`.

```python
# 1. for loop over a sequence directly
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print("Fruit:", fruit)

# 2. for loop with range(start, stop, step)
print("\nEven numbers from 2 to 10:")
for num in range(2, 11, 2):
    print(num, end=" ")
print()
```

## 6. `while-else` and `for-else` Constructs
In Python, loops can have an optional `else` block:
- The `else` block executes **if and only if the loop terminates normally** (without hitting a `break` statement).
- If the loop is exited early via `break`, the `else` block is skipped.

```python
# Example: Linear search with loop-else construct
items = [10, 20, 30, 40, 50]
target = 35

for item in items:
    if item == target:
        print(f"Found target {target} in list!")
        break
else:
    # Executes only if loop finished all iterations without finding target
    print(f"Target {target} was NOT found in list.")
```

## 7. Practice Questions

### Practice Question 1: Print Numbers from 1 to 100
Write a program using a `while` loop to print all integers from 1 to 100.

```python
# Practice 1: Print numbers from 1 to 100
num = 1
while num <= 100:
    print(num, end=" " if num % 10 != 0 else "\n")  # Print 10 numbers per line
    num += 1
```

### Practice Question 2: Print Numbers from 100 to 1
Write a program using a `while` loop to print numbers in descending order from 100 down to 1.

```python
# Practice 2: Reverse countdown from 100 down to 1
rev = 100
while rev >= 1:
    print(rev, end=" " if rev % 10 != 1 else "\n")  # Print 10 numbers per line
    rev -= 1
```

### Practice Question 3: Multiplication Table of $n$
Write a program using a `while` loop to print the multiplication table of a number $n$ from 1 to 10.

```python
# Practice 3: Multiplication table of n
n = 7                              # Multiplier number
multiplier = 1

print(f"--- Multiplication Table of {n} ---")
while multiplier <= 10:
    product = n * multiplier
    print(f"{n} x {multiplier} = {product}")
    multiplier += 1
```

### Practice Question 4: Traverse and Print Elements of a List
Traverse the list `[1, 4, 9, 16, 25, 36, 49, 64, 81, 100]` and print each square element.

```python
# Practice 4: Traversing square numbers list with while loop
nums = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
idx = 0

print("Printing list elements:")
while idx < len(nums):
    print(f"Element at index {idx}: {nums[idx]}")
    idx += 1
```

### Practice Question 5: Linear Search in a Tuple Using While Loop
Search for a number $x = 36$ in the tuple `(1, 4, 9, 16, 25, 36, 49, 64, 81, 100)` and print the index where it is found.

```python
# Practice 5: Linear search algorithm in tuple using while loop
nums_tuple = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)
x = 36                             # Search key
i = 0
found = False

while i < len(nums_tuple):
    if nums_tuple[i] == x:
        print(f"Target {x} found at index {i}!")
        found = True
        break                      # Terminate search on first match
    i += 1

if not found:
    print(f"Target {x} was not found in tuple.")
```

### Practice Question 6: Factorial of a Number ($n!$)
Compute the factorial of a number $n$ using a loop ($n! = 1 \times 2 \times \dots \times n$).

```python
# Practice 6: Factorial calculation using loop
n = 5
fact = 1
step = 1

while step <= n:
    fact *= step
    step += 1

print(f"Factorial of {n} ({n}!):", fact)  # Output: 120
```

### Practice Question 7: Sum of First $n$ Natural Numbers
Compute the sum of the first $n$ natural numbers ($1 + 2 + \dots + n$) using a loop.

```python
# Practice 7: Sum of first n natural numbers
n = 10
total_sum = 0
current = 1

while current <= n:
    total_sum += current
    current += 1

print(f"Sum of first {n} numbers:", total_sum)  # Output: 55
```

## 8. Conceptual & Interview Questions

### Q1: What is the fundamental difference between a `while` loop and a `for` loop in Python?
**Answer:** 
- A `while` loop is **condition-controlled**: it continues executing indefinitely until a boolean expression evaluates to `False`. It is preferred when the exact number of iterations is unknown in advance.
- A `for` loop is **collection/iterator-controlled**: it iterates over a predefined sequence or iterable until exhausted. It is preferred when iterating across a known range or collection.

### Q2: How do `break`, `continue`, and `pass` differ in Python?
**Answer:**
- `break`: Exits the loop entirely and transfers execution to the statement immediately following the loop.
- `continue`: Skips the remaining code inside the loop for the current iteration and jumps directly to the next iteration.
- `pass`: A null statement used as a syntactic placeholder that does nothing and allows code execution to proceed uninterrupted.

### Q3: How does the `else` clause in a Python loop work?
**Answer:** The `else` block attached to a `for` or `while` loop executes only when the loop completes all iterations naturally without encountering a `break` statement. If `break` terminates the loop, the `else` block is bypassed.

### Q4: What is the time complexity of linear search in a sequence using a loop?
**Answer:** 
- **Best Case**: $O(1)$ when the target element is at index `0`.
- **Worst / Average Case**: $O(N)$ when the target is at the end of the sequence or absent, requiring examination of all $N$ elements.
