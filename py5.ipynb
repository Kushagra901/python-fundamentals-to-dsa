## 1. `while` Loop Fundamentals
A `while` loop repeatedly executes a block of code **as long as a specified boolean condition evaluates to `True`**.

### Anatomy of a `while` Loop:
1. **Initialization**: Set up a loop counter variable (e.g. `count = 1`).
2. **Condition**: Test the counter against a stopping threshold (e.g. `count <= 5`).
3. **Body**: Execute statements inside the loop.
4. **Update / Increment**: Modify the loop counter (e.g. `count += 1`) to guarantee eventual termination.

```python
# Basic while loop demonstration: print hello 5 times
count = 1                          # 1. Initialize counter variable
while count <= 5:                  # 2. Condition check
    print("hello")                 # 3. Loop body
    count += 1                     # 4. Increment step (prevents infinite loop)

print("Loop completed. Final counter value:", count)  # Output: 6
```

## 2. Traversing Sequences & Linear Search with `while` Loop
We can traverse lists, tuples, or strings using an integer index variable.

```python
# 1. Traversing a list of numbers using while loop
nums = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
i = 0
print("Traversing list:")
while i < len(nums):
    print(nums[i])
    i += 1

# 2. Linear search in a tuple using while loop
nums_tuple = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)
x = 36
i = 0
print("\nSearching for x =", x, "in tuple:")
while i < len(nums_tuple):
    if nums_tuple[i] == x:
        print("found the idx", i)
        break
    else:
        print("finding..")
    i += 1
```

## 3. Loop Flow Control: `break` and `continue`
- `break`: Immediately **terminates the entire loop** and transfers control to the statement directly following the loop.
- `continue`: Immediately **skips the rest of the current iteration** and jumps to the next iteration / condition check.

```python
# 1. Demonstration of break: exit loop when i reaches 3
print("--- Demonstrating break ---")
i = 1
while i <= 5:
    print(i)
    if i == 3:
        break                      # Terminate loop early
    i += 1
print("end of loop")

# 2. Demonstration of continue: skip iteration when i is 3
print("\n--- Demonstrating continue ---")
i = 0
while i <= 5:
    if i == 3:
        i += 1
        continue                   # Skip rest of body for i = 3
    print(i)
    i += 1
```

## 4. `for` Loops in Python
A `for` loop is used for sequential traversal (iterating over lists, tuples, strings, or ranges).
Python also supports an optional `else` block on loops, which runs when the loop finishes all iterations without hitting a `break`.

```python
# 1. for loop on list
list_nums = [1, 2, 3, 4, 5, 6]
veggie = ["potaoto", "tomato"]
print("Iterating over list:")
for vals in veggie:
    print(vals)

# 2. for loop on tuple
tup = (1, 2, 4, 5)
print("\nIterating over tuple:")
for num in tup:
    print(num)

# 3. for loop on string with break and optional else
str_val = "kuosh"
print("\nIterating over string with search:")
for char in str_val:
    if char == "o":
        print("o found")
        break
    print(char)
else:
    print("end")
```

## 5. The `range()` Function
The `range()` function returns a sequence of numbers:
- `range(stop)`: From `0` up to `stop - 1`
- `range(start, stop)`: From `start` up to `stop - 1`
- `range(start, stop, step)`: Increments by `step`

```python
# 1. Indexing into a range sequence
seq = range(5)
print("seq[0]:", seq[0])
print("seq[1]:", seq[1])
print("seq[2]:", seq[2])
print("seq[3]:", seq[3])

# 2. range(10) -> stopping condition at 10 (0 to 9)
print("\nrange(10):")
for i in range(10):
    print(i, end=" ")
print()

# 3. range(2, 10) -> start at 2, stop before 10
print("\nrange(2, 10):")
for i in range(2, 10):
    print(i, end=" ")
print()

# 4. range(2, 10, 2) -> start at 2, stop before 10, step by 2
print("\nrange(2, 10, 2):")
for i in range(2, 10, 2):
    print(i, end=" ")
print()
```

## 6. The `pass` Statement
`pass` is a null statement used as a syntactic placeholder when a statement is required syntactically but no code needs to execute.

```python
# 1. Using pass in a loop
for i in range(5):
    # empty block placeholder
    pass
print("some work")

# 2. Using pass in conditionals
for i in range(5):
    pass
if i > 5:
    pass
print("some useful work")
```

Search for a number `x = 49` in the tuple `(1, 4, 9, 16, 25, 36, 49, 64, 81, 100)` and print the index where it is found.

```python
# Practice 5: Search in tuple using for loop
nums = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)
x = 49
i = 0
for el in nums:
    if el == x:
        print("number found", i)
        break
    i += 1
```

### Practice Question 6: Sum of First $n$ Numbers
Calculate the sum of the first $n$ numbers using both a `for` loop and a `while` loop.

```python
# Practice 6A: Sum using for loop
n = 8
sum_val = 0
for i in range(1, n + 1):
    sum_val += i
print("Sum of first", n, "numbers (for loop):", sum_val)

# Practice 6B: Sum using while loop
n = 8
sum_val = 0
i = 1
while i <= n:
    sum_val += i
    i += 1
print("Sum of first", n, "numbers (while loop):", sum_val)
```

### Practice Question 7: Factorial of a Number
Calculate the factorial of a number ($n!$) using both a `while` loop and a `for` loop.

```python
# Practice 7A: Factorial using while loop
n = 5
fact = 1
i = 1
while i <= n:
    fact *= i
    i += 1
print("Factorial of", n, "(while loop):", fact)

# Practice 7B: Factorial using for loop
n = 5
fact = 1
for i in range(1, n + 1):
    fact *= i
print("Factorial of", n, "(for loop):", fact)
```

## 8. Conceptual & Interview Questions

### Q1: What is the fundamental difference between a `while` loop and a `for` loop in Python?
**Answer:** 
- A `while` loop is **condition-controlled**: it continues executing indefinitely until a boolean expression evaluates to `False`. It is preferred when the exact number of iterations is unknown in advance.
- A `for` loop is **collection/iterator-controlled**: it iterates over a predefined sequence or iterable until exhausted. It is preferred when iterating across a known range or collection.

### Q2: How do `break`, `continue`, and `pass` differ in Python?
**Answer:**
- `break`: Exits the loop entirely and transfers execution to the statement immediately following the loop.
- `continue`: Skips the remaining code inside the loop for the current iteration and jumps directly to the next iteration.
- `pass`: A null statement used as a syntactic placeholder that does nothing and allows code execution to proceed uninterrupted.

### Q3: How does the `else` clause in a Python loop work?
**Answer:** The `else` block attached to a `for` or `while` loop executes only when the loop completes all iterations naturally without encountering a `break` statement. If `break` terminates the loop, the `else` block is bypassed.

### Q4: What is the time complexity of linear search in a sequence using a loop?
**Answer:** 
- **Best Case**: $O(1)$ when the target element is at index `0`.
- **Worst / Average Case**: $O(N)$ when the target is at the end of the sequence or absent, requiring examination of all $N$ elements.
