## 1. Opening, Reading & Closing Files
Python utilizes the built-in `open(filename, mode)` function to access files stored on disk.

### Common File Modes:
| Mode | Name | Description |
| :--- | :--- | :--- |
| `'r'` | Read (Default) | Opens for reading; raises `FileNotFoundError` if the file does not exist. |
| `'w'` | Write | Opens for writing; creates the file if missing, **truncates (overwrites)** if existing. |
| `'a'` | Append | Opens for writing; creates the file if missing, appends data to the end without truncating. |
| `'r+'` | Read + Write | Reads and writes without truncating; pointer starts at index `0`. |
| `'w+'` | Write + Read | Truncates existing file (or creates new), allows reading and writing. |

```python
# 1. Creating a sample file for demonstration
f = open("demo.txt", "w")
f.write("Hello Python Learners\nWe are mastering File I/O\nPython is powerful.")
f.close()                           # Always close explicitly when not using 'with'

# 2. Reading all contents using .read()
f = open("demo.txt", "r")
content = f.read()                  # Reads entire file content as a single string
print("--- Entire File Content ---")
print(content)
f.close()

# 3. Reading fixed characters using .read(n)
f = open("demo.txt", "r")
first_5 = f.read(5)                 # Reads first 5 characters
print("\nFirst 5 characters:", repr(first_5))
f.close()

# 4. Reading line-by-line using .readline()
f = open("demo.txt", "r")
line1 = f.readline()                # Reads first line including '\n'
line2 = f.readline()                # Reads second line
print("\nLine 1:", line1.strip())
print("Line 2:", line2.strip())
f.close()
```

## 2. Writing & Appending to Files
- Mode `'w'` truncates (clears) existing content prior to writing.
- Mode `'a'` appends new data to the end of the file, preserving earlier content.

```python
# Appending to an existing file
f = open("demo.txt", "a")
f.write("\nNext topic: Object-Oriented Programming")
f.close()

# Verifying updated content
f = open("demo.txt", "r")
print(f.read())
f.close()
```

## 3. Safe File Handling with Context Manager (`with` Syntax)
The `with` statement encapsulates file operations within a Python context manager.
It **automatically closes the file descriptor**, even if an unhandled exception or early `return` occurs.

```python
# Reading with context manager
with open("demo.txt", "r") as f:
    data = f.read()
    print("Read via context manager:")
    print(data)

# Writing with context manager
with open("sample.txt", "w") as f:
    f.write("Sample configuration and logs.")

# File is automatically closed here; checking f.closed verifies this
print("Is file closed?", f.closed)   # Output: True
```

## 4. File Deletion using `os` Module
To delete files from disk programmatically, import the standard `os` module and use `os.remove(path)`.

```python
import os

# Check if sample.txt exists before deleting
if os.path.exists("sample.txt"):
    os.remove("sample.txt")
    print("sample.txt successfully removed.")
else:
    print("File does not exist.")
```

## 5. Practice Questions (File I/O)

### Practice Question 1: Find and Replace in a File
Create a file `practice.txt`, write sample text containing the word `"java"`, and replace every occurrence of `"java"` with `"python"`.

```python
# Step 1: Create practice.txt with initial content
with open("practice.txt", "w") as f:
    f.write("hi everyone\nwe are learning file i/o\nusing java\ni like programming in java.")

# Step 2: Read and replace word
with open("practice.txt", "r") as f:
    data = f.read()

new_data = data.replace("java", "python")

# Step 3: Overwrite with modified data
with open("practice.txt", "w") as f:
    f.write(new_data)

# Step 4: Verify the updated content
with open("practice.txt", "r") as f:
    print("Updated practice.txt content:")
    print(f.read())
```

### Practice Question 2: Word Search & Line Locator
Write functions `check_for_word(word)` and `check_for_line(word)` to search for a target word inside `practice.txt`.

```python
# 1. Function to check if a word exists in file
def check_for_word(word, filename="practice.txt"):
    with open(filename, "r") as f:
        data = f.read()
        if word in data:
            print(f"Word '{word}' found in file.")
            return True
        else:
            print(f"Word '{word}' not found in file.")
            return False

# 2. Function to find the 1-indexed line number containing the target word
def check_for_line(word, filename="practice.txt"):
    line_no = 1
    with open(filename, "r") as f:
        for line in f:
            if word in line:
                return line_no
            line_no += 1
    return -1                       # Return -1 if word not found

check_for_word("python")
line_idx = check_for_line("python")
print("Word 'python' first occurs on line:", line_idx)
```

### Practice Question 3: Parse Comma-Separated Numbers & Count Evens
Read a file containing integers separated by commas (e.g., `1,2,76,84,100,111`) and compute the total count of even numbers.

```python
# Write comma-separated numbers into practice.txt
with open("practice.txt", "w") as f:
    f.write("1,2,76,84,100,111")

# Read, parse, and count even numbers
count_even = 0
with open("practice.txt", "r") as f:
    data = f.read()
    nums = data.split(",")
    for val in nums:
        if int(val.strip()) % 2 == 0:
            count_even += 1

print("Parsed values:", nums)
print("Count of even numbers:", count_even)  # Output: 4 (2, 76, 84, 100)
```

## 6. Conceptual & Technical Interview Questions

### Q1: Why is the `with open(...) as f:` syntax preferred over manual `open()` and `close()`?
**Answer:**
- The `with` statement utilizes Python's context management protocol (`__enter__` and `__exit__`).
- It guarantees that the file descriptor is properly flushed and closed immediately when the block exits, even if an unhandled exception or return statement interrupts execution, preventing resource and memory leaks.

### Q2: What is the difference between file access modes `'r+'` and `'w+'`?
**Answer:**
- `'r+'` opens an existing file for both reading and writing without truncating it; the file pointer starts at byte 0. If the file does not exist, it raises a `FileNotFoundError`.
- `'w+'` opens a file for both writing and reading. If the file exists, it **immediately truncates (erases)** the content to 0 bytes; if the file does not exist, it creates a new empty file.

### Q3: How do `read()`, `readline()`, and `readlines()` compare in memory consumption?
**Answer:**
- `read()` loads the entire file into memory as a single string ($O(N)$ memory).
- `readlines()` reads all lines into memory as a list of strings ($O(N)$ memory).
- `readline()` reads only one line at a time into memory ($O(\text{line\_length})$ memory).
- For massive multi-gigabyte files, iterating directly over the file object (`for line in f:`) is the most memory-efficient approach as it streams one line at a time via a generator buffer.

### Q4: How do `tell()` and `seek()` manipulate the file pointer in Python?
**Answer:**
- `f.tell()` returns the current integer byte offset of the file pointer.
- `f.seek(offset, whence)` relocates the file pointer to `offset` bytes relative to `whence` (0 = start of file, 1 = current position, 2 = end of file).
