## 1. Lists in Python
A **list** is an ordered, **mutable** collection that can store elements of different data types (`int`, `float`, `str`, etc.) in square brackets `[]`.


```python
# 1. List creation and heterogeneous data storage
marks = [12, 3, 4, 5656, 5765, 43, 234, 45]
print("Type of marks:", type(marks))
print("Marks list:", marks)
print("marks[3]:", marks[3])
print("marks[2]:", marks[2])
print("Length of marks:", len(marks))

# Storing elements of different types (int, float, str) all in one list
student = ["kararn", 456, 213, "mumbai"]
print("Student list:", student)

# 2. Lists are MUTABLE (elements can be changed in-place)
print("Original student[0]:", student[0])
student[0] = "arjun"                   # Modify index 0
print("Modified student list:", student)

# 3. Slicing in lists (ending index is not included)
marks_slice = [86, 57, 98, 54, 678, 98]
print("Slice [1:4]:", marks_slice[1:4])     # [57, 98, 54]
print("Slice [-3:-1]:", marks_slice[-3:-1]) # [54, 678]
```

## 2. Essential List Methods
Built-in list methods for adding, ordering, and removing elements:
- `.append(val)`: Adds an item to the end of the list.
- `.sort()`: Sorts the list in ascending order in-place.
- `.sort(reverse=True)`: Sorts in descending order.
- `.reverse()`: Reverses the list elements in-place.
- `.insert(index, val)`: Inserts element at specified index.
- `.remove(val)`: Removes the first occurrence of value.
- `.pop(index)`: Removes and returns element at specified index.


```python
# List methods demonstration

# 1. append(val)
m = [1, 2, 3, 4]
m.append(5)
print("After append(5):", m)

# 2. sort() and sort(reverse=True)
student_scores = [2, 8, 9, 5, 6]
words = ["ba", "li", "ap"]
words.sort()
print("Sorted strings:", words)        # ['ap', 'ba', 'li']

student_scores.sort()
print("Sorted ascending:", student_scores)

student_scores.sort(reverse=True)
print("Sorted descending:", student_scores)

# 3. reverse()
marks1 = [4, 5, 6, 2, 3]
marks1.reverse()
print("Reversed list:", marks1)

# 4. insert(index, val)
marks_ins = [1, 2, 4, 5]
marks_ins.insert(2, 3)                 # Insert value 3 at index 2
print("After insert(2, 3):", marks_ins)

# 5. remove(val) - removes first occurrence
marks_rem = [1, 2, 1, 4, 5]
marks_rem.remove(1)
print("After remove(1):", marks_rem)

# 6. pop(index) - removes element at index
marks_pop = [1, 2, 3, 4]
popped_item = marks_pop.pop(2)
print("Popped item at index 2:", popped_item)
print("List after pop:", marks_pop)
```

## 3. Tuples in Python
A **tuple** is an ordered, **immutable** sequence created with parentheses `()`. Once created, tuple values cannot be modified.


```python
# Tuple creation and indexing
mark_tup = (1, 2, 3, 4)
print("Tuple:", mark_tup)
print("Element at index 1:", mark_tup[1])
print("Element at index 3:", mark_tup[3])
print("Type:", type(mark_tup))

# Slicing in tuple
print("Tuple slice [1:3]:", mark_tup[1:3]) # (2, 3)

# Tuple Methods
# 1. index(val): Returns first index of value
pri = (1, 2, 3)
print("pri.index(2):", pri.index(2))       # 1

# 2. count(val): Returns frequency of value in tuple
pp = (1, 2, 3, 1, 1, 3, 2, 4, 3, 3)
print("pp.count(1):", pp.count(1))         # 3
```

## 4. Palindrome Verification Algorithms
A sequence is a palindrome if it reads the same forwards and backwards.


```python
# Method 1: Two-Pointer Technique (O(N) time, O(1) space)
arr = [1, 2, 3, 2, 1]
n = len(arr)
l = 0
r = n - 1
is_palindrome = True

while l < r:
    if arr[l] != arr[r]:
        is_palindrome = False
        print("No, it is not a palindrome")
        break
    l += 1
    r -= 1
else:
    print("True: It is a palindrome (Two-pointer method)")

# Method 2: Reverse and Copy Method (O(N) time, O(N) space)
list1 = [1, 2, 1]
copy_list1 = list1.copy()
copy_list1.reverse()

if copy_list1 == list1:
    print("Palindrome (Copy-reverse method)")
else:
    print("Not a palindrome")
```

## 5. Algorithmic Problem: Interval / Window Merging (`solution`)
Given a list of interval windows `[start, end]`, merge overlapping intervals in $O(N \log N)$ time.


```python
def solution(windows):
    """
    Merges overlapping intervals.
    
    Parameters:
        windows (list of list of int): List of intervals [start, end]
        
    Returns:
        list of list of int: Merged non-overlapping intervals
    """
    if not windows:
        return []

    # Sort intervals by starting boundary
    windows.sort(key=lambda x: x[0])
    
    # Initialize merged list with the first interval as a 2D list
    merged = [[windows[0][0], windows[0][1]]]

    # Iterate over remaining intervals
    for start, end in windows[1:]:
        if start <= merged[-1][1]:
            # Overlapping intervals: merge by taking maximum end time
            merged[-1][1] = max(merged[-1][1], end)
        else:
            # Non-overlapping interval: append to result list
            merged.append([start, end])

    return merged

# Test example
test_windows = [[1, 3], [2, 6], [8, 10], [15, 18]]
print("Input intervals:", test_windows)
print("Merged intervals:", solution(test_windows))  # Expected: [[1, 6], [8, 10], [15, 18]]
```

## 7. Conceptual & Interview Questions

### Q1: What is the key difference between a list and a tuple?
**Answer:** 
- **Lists** are mutable (elements can be added/modified in-place), unhashable, and use dynamic array over-allocation.
- **Tuples** are immutable (cannot be altered after creation), consume less memory, and are hashable (can be used as dictionary keys).

### Q2: What is the time complexity of `solution(windows)`?
**Answer:** $O(N \log N)$ where $N$ is the number of intervals, dominated by `windows.sort()`. The merging pass runs in linear $O(N)$ time.

### Q3: Why is the Two-Pointer palindrome check more memory-efficient than `list.copy()` + `reverse()`?
**Answer:** The Two-Pointer method inspects elements in-place with $O(1)$ extra space, whereas `copy()` creates a new duplicate list in memory requiring $O(N)$ auxiliary space.
