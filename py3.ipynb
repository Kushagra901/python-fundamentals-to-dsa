## 1. Lists in Python
A **list** is an ordered, mutable sequence of elements. Lists can store heterogeneous data types (`int`, `float`, `str`, objects) in a single collection.


```python
# 1. List creation with heterogeneous data types
student = ["karan", 456, 213.5, "mumbai"]
print("Student list:", student)
print("Type:", type(student))
print("First element (index 0):", student[0])
print("Length of list:", len(student))

# 2. Lists are MUTABLE (elements can be modified in-place)
student[0] = "arjun"  # Replace "karan" with "arjun"
print("Modified student list:", student)

# 3. List slicing [start : stop]
marks = [86, 57, 98, 54, 678, 98]
print("Slice [1:4]:", marks[1:4])   # [57, 98, 54]
print("Slice [-3:-1]:", marks[-3:-1]) # [54, 678]

# 4. Essential List Methods
sample_list = [4, 2, 8, 5, 1]

# append(val): Adds an element to the end of the list - O(1) amortized
sample_list.append(10)
print("After append(10):", sample_list)

# sort(): Sorts the list in ascending order in-place - O(N log N)
sample_list.sort()
print("After sort():", sample_list)

# sort(reverse=True): Sorts the list in descending order in-place
sample_list.sort(reverse=True)
print("After sort(reverse=True):", sample_list)

# reverse(): Reverses the list elements in-place - O(N)
sample_list.reverse()
print("After reverse():", sample_list)

# insert(index, val): Inserts value at specified index - O(N)
sample_list.insert(2, 99)
print("After insert(2, 99):", sample_list)

# remove(val): Removes the first occurrence of val - O(N)
sample_list.remove(99)
print("After remove(99):", sample_list)

# pop(index): Removes and returns element at specified index - O(N) (or O(1) for last element)
popped_val = sample_list.pop(2)
print(f"Popped value: {popped_val} | Remaining list: {sample_list}")
```

## 2. Tuples in Python
A **tuple** is an ordered, **immutable** sequence of elements enclosed in parentheses `()`. Once created, tuple elements cannot be modified, inserted, or removed.


```python
# 1. Tuple creation and indexing
numbers_tuple = (1, 2, 3, 4, 2, 2)
print("Tuple:", numbers_tuple)
print("Type:", type(numbers_tuple))
print("Element at index 1:", numbers_tuple[1])

# 2. Tuple slicing
print("Tuple slice [1:3]:", numbers_tuple[1:3]) # (2, 3)

# 3. Tuple Methods
# index(val): Returns first index of val
print("Index of 3:", numbers_tuple.index(3))     # 2

# count(val): Returns total occurrences of val
print("Count of 2:", numbers_tuple.count(2))     # 3
```

## 3. Palindrome Verification Algorithms
A sequence is a **palindrome** if it reads identically forwards and backwards (e.g. `[1, 2, 3, 2, 1]` or `"racecar"`).


```python
# Method 1: Two-Pointer Technique (Optimal: O(N) Time, O(1) Extra Space)
def is_palindrome_two_pointer(arr):
    left = 0
    right = len(arr) - 1
    
    while left < right:
        if arr[left] != arr[right]:
            return False  # Mismatch found
        left += 1
        right -= 1
        
    return True  # All symmetric pairs matched

print("Two-pointer test on [1, 2, 1]:", is_palindrome_two_pointer([1, 2, 1]))  # True
print("Two-pointer test on [1, 2, 3]:", is_palindrome_two_pointer([1, 2, 3]))  # False

# Method 2: Reverse-and-Compare (O(N) Time, O(N) Extra Space)
list1 = [1, 2, 1]
copy_list1 = list1.copy()  # Create shallow copy
copy_list1.reverse()       # Reverse copy in-place

if copy_list1 == list1:
    print("List is a palindrome (Reverse-and-compare method)")
else:
    print("List is NOT a palindrome")
```

## 4. Algorithmic Problem: Interval / Window Merging (`solution`)
**Problem Statement:** Given a collection of intervals/windows `[start, end]`, merge all overlapping intervals and return a list of mutually exclusive intervals sorted by start coordinate.

**Algorithm:**
1. If the input is empty, return `[]`.
2. Sort the intervals based on their starting coordinate (`x[0]`) in $O(N \log N)$ time.
3. Initialize the `merged` list with the first interval as a 2D list: `[[windows[0][0], windows[0][1]]]`.
4. Iterate through each subsequent interval `[start, end]`:
   - If `start <= merged[-1][1]` (overlapping), extend the previous interval's end: `merged[-1][1] = max(merged[-1][1], end)`.
   - Else (no overlap), append `[start, end]` as a new interval to `merged`.
5. Return the `merged` list.


```python
def solution(windows):
    """
    Merges overlapping interval windows.
    
    Parameters:
        windows (list of list of int): List of intervals [start, end]
        
    Returns:
        list of list of int: Merged non-overlapping intervals
        
    Time Complexity: O(N log N) due to initial sorting
    Space Complexity: O(N) to store merged output
    """
    if not windows:
        return []

    # Sort intervals by starting boundary
    windows.sort(key=lambda x: x[0])
    
    # Initialize merged list with the first interval
    merged = [[windows[0][0], windows[0][1]]]

    # Iterate over remaining intervals
    for start, end in windows[1:]:
        if start <= merged[-1][1]:
            # Overlapping intervals: expand the end time
            merged[-1][1] = max(merged[-1][1], end)
        else:
            # Non-overlapping: append new interval
            merged.append([start, end])

    return merged

# Test cases
test_intervals = [[1, 3], [2, 6], [8, 10], [15, 18]]
result = solution(test_intervals)
print("Input Intervals:  [[1, 3], [2, 6], [8, 10], [15, 18]]")
print("Merged Intervals:", result)  # Expected: [[1, 6], [8, 10], [15, 18]]
```

## 6. Interview Questions & Conceptual Insights

### Q1: What is the primary difference between lists and tuples in Python?
**Answer:** 
- **Lists** are mutable (dynamic arrays with over-allocation for fast append) and unhashable.
- **Tuples** are immutable (fixed-size contiguous blocks of memory), consume less memory, and are hashable (can be used as dictionary keys if elements are hashable).

### Q2: Why is the Two-Pointer palindrome check preferred over reverse copying?
**Answer:** The two-pointer approach checks elements in-place with $O(1)$ auxiliary space and can terminate early on the first mismatch. Reverse copying always allocates $O(N)$ extra memory and requires full list reversal before comparison.

### Q3: What is the time complexity of interval merging?
**Answer:** $O(N \log N)$, dominated by sorting the $N$ intervals. The subsequent merging loop takes linear $O(N)$ time with a single pass through the list.
