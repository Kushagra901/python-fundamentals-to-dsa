## 1. Deleting Attributes & Objects (`del` Keyword)
- The `del` keyword removes variable bindings from a namespace or deletes attributes from an object.
- **Deleting an Attribute**: `del obj.attribute` removes the key from `obj.__dict__`. Accessing it afterwards raises an `AttributeError`.
- **Deleting an Object**: `del obj` unbinds the name reference and decrements the object's reference counter.

### Syntax:
```python
del s1.name    # Removes attribute 'name' from instance s1
del s1         # Removes object reference s1
```

## 2. Public vs. Private Attributes & Methods
- **Public Members**: Accessible from anywhere inside and outside the class. (Default in Python).
- **Private Members**: Prefixed with double leading underscores (e.g., `__acc_pass`, `__hello()`).
- **Name Mangling**: Python internally rewrites `__variable` to `_ClassName__variable` to prevent accidental overwrites in subclasses.
- Private attributes and methods should be accessed and manipulated via public interface methods.

## 3. Inheritance in Python (Single, Multi-Level & Multiple)
Inheritance enables a derived (child) class to inherit attributes and methods from one or more base (parent) classes, promoting DRY (Don't Repeat Yourself) architecture.

### Inheritance Types:
1. **Single Inheritance**: Child class inherits from a single parent class (`Parent -> Child`).
2. **Multi-level Inheritance**: Child class inherits from a derived class forming a chain (`Grandparent -> Parent -> Child`).
3. **Multiple Inheritance**: Child class inherits directly from two or more parent classes (`ParentA, ParentB -> Child`).

### Method Resolution Order (MRO):
- In multiple inheritance, Python resolves attribute and method lookups using the C3 Linearization algorithm.
- For `class C(A, B):`, lookup proceeds: `C -> A -> B -> object`.

## 4. The `super()` Method
- The `super()` built-in function returns a proxy object that delegates method and attribute lookups to a parent (or sibling) class in the inheritance chain.
- **Constructor Chaining**: Most frequently used in derived class constructors via `super().__init__(*args)` to execute parent initialization logic without hardcoding parent class names.
- **Dynamic MRO Resolution**: Automatically follows the class's Method Resolution Order (MRO), ensuring clean cooperative multiple inheritance without multiple calls to shared base classes.

### Syntax:
```python
class Parent:
    def __init__(self, attr):
        self.attr = attr

class Child(Parent):
    def __init__(self, attr, child_attr):
        super().__init__(attr)      # Calls Parent.__init__
        self.child_attr = child_attr
```

## 5. Class Methods (`@classmethod` Decorator)
- **Class Method**: A method bound to the class itself rather than to an individual instance. It receives the class (`cls`) as its first parameter instead of `self`.
- **`@classmethod` Decorator**: Used to designate a method as a class method.
- **Modifying Class State**: Used when you need to access or modify class-level attributes that apply across all instances.
- **Factory Methods / Alternative Constructors**: Frequently utilized in Python to provide alternative ways of creating objects (e.g., parsing from strings, JSON, or configuration dictionaries).

### Syntax:
```python
class Person:
    species = "Homo sapiens"

    @classmethod
    def change_species(cls, new_species):
        cls.species = new_species   # Directly updates class attribute
```

## 6. The `@property` Decorator (Getters & Computed Attributes)
- **`@property` Decorator**: Converts a method into a read-only getter attribute, allowing it to be accessed via dot notation (`obj.attribute`) without function call parentheses `()`.
- **Dynamic / Computed Attributes**: Automatically recalculates derived values on the fly whenever accessed, keeping data consistent when underlying attributes mutate.
- **Encapsulation & Backward Compatibility**: Allows refactoring an attribute from a simple stored value into a computed property without breaking existing code that accesses `obj.attribute`.

### Syntax:
```python
class Student:
    def __init__(self, phy, chem, math):
        self.phy = phy
        self.chem = chem
        self.math = math

    @property
    def percentage(self):
        return (self.phy + self.chem + self.math) / 3
```

## 7. Polymorphism & Operator Overloading
- **Polymorphism** ("many forms"): The capability of the same operator or method signature to exhibit different behaviors when applied to different classes or data types.
  - Built-in example: The `+` operator adds two numbers (`1 + 3 = 4`) but concatenates two strings (`"a" + "b" = "ab"`).
- **Operator Overloading**: Python achieves operator polymorphism using special double-underscore methods ("dunder" / magic methods).
- When Python evaluates an expression like `a + b`, it internally invokes `a.__add__(b)`.
- By implementing dunder methods in custom classes, user-defined objects can interact seamlessly using standard Python operators (`+`, `-`, `*`, `>`, `<`, `==`, etc.).

### Common Operator Dunder Methods:
| Operator | Dunder Method | Operation |
| :--- | :--- | :--- |
| `a + b` | `__add__(self, other)` | Addition |
| `a - b` | `__sub__(self, other)` | Subtraction |
| `a * b` | `__mul__(self, other)` | Multiplication |
| `a / b` | `__truediv__(self, other)` | True Division |
| `a // b` | `__floordiv__(self, other)` | Floor Division |
| `a % b` | `__mod__(self, other)` | Modulo / Remainder |
| `a > b` | `__gt__(self, other)` | Greater Than |
| `a < b` | `__lt__(self, other)` | Less Than |
| `a == b` | `__eq__(self, other)` | Equality Check |

## 9. Conceptual & Technical Interview Questions

### Q1: How does the `del` keyword work in Python, and does it immediately free memory?
**Answer:**
- `del` unbinds a variable name from an object or deletes an attribute/key from an object's dictionary.
- It does **not** directly free memory; it decrements the object's reference count by 1.
- Memory is deallocated only when the object's reference count drops to zero, at which point Python's garbage collector frees the block (calling `__del__` if defined).

### Q2: Does Python support truly private attributes or methods?
**Answer:**
- No. Python does not provide true hardware- or compiler-enforced access restrictions like `private` in C++ or Java.
- Instead, Python employs **name mangling**: any identifier prefixed with double underscores (e.g., `__name`) is rewritten to `_ClassName__name`.
- While `p1.__name` raises an `AttributeError`, the attribute can still be accessed via `p1._Person__name`. Privacy in Python relies on developer convention ("we are all consenting adults here").

### Q3: What is the difference between Single, Multi-level, and Multiple Inheritance?
**Answer:**
- **Single Inheritance**: One child class inherits from one parent class (`class Child(Parent):`).
- **Multi-level Inheritance**: A class inherits from a class that already inherits from another class (`class GrandChild(Child):` where `Child` inherits from `Parent`).
- **Multiple Inheritance**: A class inherits directly from two or more parent classes (`class Child(ParentA, ParentB):`), combining features from multiple lineages.

### Q4: What is Method Resolution Order (MRO) and how does Python resolve method calls in multiple inheritance?
**Answer:**
- Method Resolution Order (MRO) is the sequence in which Python searches base classes when looking up methods and attributes.
- Python uses the **C3 Linearization algorithm** to guarantee monotonicity and prevent ambiguous lookups (such as the Diamond Problem).
- For `class C(A, B):`, MRO searches `C` first, then `A`, then `B`, and finally the universal base class `object`.
- The MRO sequence can be inspected on any class using `ClassName.__mro__` or `ClassName.mro()`.

### Q5: What is the purpose of the `super()` method and why is it preferred over calling `Parent.__init__(self)` directly?
**Answer:**
- `super()` returns a proxy delegate that dynamically resolves methods based on the current class's Method Resolution Order (MRO).
- **Decoupling**: Avoids hardcoding the parent class name in derived classes, making code more maintainable during refactoring or class renaming.
- **Cooperative Multiple Inheritance**: In diamond-shaped multiple inheritance hierarchies, calling parent methods directly (e.g., `ParentA.__init__(self)`) can cause shared ancestor constructors to execute multiple times. `super()` guarantees that each class in the MRO chain is visited exactly once.

### Q6: What is the difference between Instance Methods, Class Methods (`@classmethod`), and Static Methods (`@staticmethod`)?
**Answer:**
- **Instance Method**: Receives `self` (the instance) as its first argument. Can inspect and mutate instance-level state (`self.attr`) as well as class-level state.
- **Class Method (`@classmethod`)**: Receives `cls` (the class) as its first argument. Can access and modify class-level state (`cls.attr`) shared across all instances, but cannot access instance attributes. Commonly used as factory methods / alternative constructors.
- **Static Method (`@staticmethod`)**: Receives neither `self` nor `cls`. Behaves like a standard utility function scoped within the class's namespace for logical grouping. Cannot modify instance or class state.

### Q7: What is the `@property` decorator in Python and when should you use it?
**Answer:**
- `@property` is a built-in decorator used to define a getter method that can be accessed like an attribute (i.e., `obj.prop` without parentheses).
- **Getters & Setters**: Provides a pythonic way to implement getters, setters (`@prop.setter`), and deleters (`@prop.deleter`) while enforcing encapsulation and data validation.
- **Dynamic / Computed Attributes**: Ideal for attributes whose values depend on other mutable attributes (e.g., calculating percentage from subject marks), ensuring values are always computed on-the-fly and never stale.
- **API Stability**: Allows refactoring a public attribute into a computed property without breaking external code or changing public interfaces.

### Q8: What is Operator Overloading and how do special (dunder) methods enable Polymorphism in Python?
**Answer:**
- **Operator Overloading**: Giving custom definitions to standard Python operators (`+`, `-`, `*`, `>`, `<`, etc.) so that they operate intuitively on user-defined objects.
- **Dunder Methods**: Python translates operator syntax into internal method invocations (e.g., `a + b` calls `a.__add__(b)`, `a > b` calls `a.__gt__(b)`).
- **Polymorphic Behavior**: By implementing these dunder methods in custom classes, distinct classes can respond to the same operator in a type-appropriate manner, enabling custom classes to seamlessly integrate with Python's built-in syntax, functions (like `len()`, `str()`), and idioms.
